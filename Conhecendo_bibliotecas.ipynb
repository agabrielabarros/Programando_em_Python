O método json_normalize() é uma função da biblioteca Pandas utilizada para transformar dados em formato JSON em um DataFrame. Ela é muito útil quando se trabalha com dados em JSON, pois muitas vezes eles estão em formato aninhado e precisam ser normalizados para que possam ser trabalhados facilmente.

> Vamos deixar essa coluna de 'contas' em um formato normal com normalize. Cada chave dessa veio como colunas, de forma separada

```python
pd.json_normalize(dados_churn['conta']).head()
```

> Vamos fazer isso para com mais uma coluna que estão com os dados juntos.

```python
pd.json_normalize(dados_churn['telefone']).head()
```

Mas fazer isso uma por uma dá muito trabalho... Se quisermos um dataframe totalmente normalizado em uma tabela que possamos trabalhar é possível fazer tudo de uma vez só, ao invés de uma a uma.

## Transformando dados em uma tabela

> Podemos aplicar o método json_normalize() em um dataframe (como fizemos acima) ou em um objeto JSON. A principal diferença é que, ao normalizar uma coluna específica, aplicamos o método no dataframe. Já ao normalizar todo o JSON de uma vez, aplicamos o método no objeto JSON.

* Em um DataFrame ⇾ normaliza apenas uma coluna

* Em um objeto JSON ⇾ normaliza todas as colunas aninhadas de uma só vez

```python
import json
```

```python
with open('/content/drive/MyDrive/Colab Notebooks/Limpeza de dados Json/dataset-telecon.json') as f:
  json_bruto = json.load(f)
```

```python
json_bruto
```

```python
dados_normalizados = pd.json_normalize(json_bruto)
dados_normalizados.head()
```

Exatamente! O **with** é utilizado para abrir o arquivo JSON e carregar o conteúdo de forma segura e eficiente, garantindo que o arquivo será fechado após o uso. Abrimos o arquivo “dados.json” e usamos o método *json.load()*. O método **json.load()** é uma função Python que permite carregar um arquivo JSON em memória e convertê-lo em um objeto Python que pode ser usado no programa. Logo após, inserimos o arquivo carregado no método **json_normalize** que converte o arquivo JSON em uma tabela relacional, em que as chaves do objeto JSON são transformadas em colunas e seus valores são adicionados nas linhas correspondentes.

Agora sim nosso dataframe está pronto para começarmos a **tratar**.

O que devemos fazer quando temos colunas com tipos incorretos no banco de dados — principalmente quando pretendemos inserir estes dados em um modelo de Machine Learning? Transformá-los na tipagem correta por meio de um cast, permitindo que este modelo possa tratá-los de forma correta e gerar resultados assertivos.

## Modificando o tipo da coluna

Identificando valores estranhos
Para realizar o cast, acessaremos a célula vazia localizada embaixo da tabela que foi retornada por nosso último comando.

Nela, escreveremos o nome do campo de dados (dados_normalizados[]), que terá entre colchetes e aspas simples o nome da coluna a ser corrigida (conta.cobranca.Total).

À direita dos colchetes, adicionaremos o sinal de igual e a operação que queremos realizar — neste caso, queremos retornar a mesma coluna, só que com o tipo diferente, portanto copiaremos todo o conteúdo à esquerda do sinal de igual e colaremos à sua direita, adicionando o método .astype(), informando entre parênteses o novo tipo de dado: float.

```python
dados_normalizados['conta.cobranca.Total'] = dados_normalizados['conta.cobranca.Total'].astype(float)
```

Isto significa que temos uma string estranha nesta coluna. Por este motivo, ela adquiriu o tipo object.
